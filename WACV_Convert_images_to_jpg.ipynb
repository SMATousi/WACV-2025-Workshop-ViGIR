```python
from PIL import Image
import os
from tqdm import tqdm
```

```python
# List input and output path 
input_folder = "/root/home/data/"
output_path = "/root/home/data_jpg/"

# Create output folder based on output path
if not os.path.exists(output_path):
    os.makedirs(output_path)

# Batch convert all .tif files in a directory
for file in tqdm(os.listdir(input_folder)):
    if file.endswith('.tif'):
        image = Image.open(os.path.join(input_folder, file))
        output_file = file.replace('.tif', '.jpg')
        #print("Converting img: {}".format(file))
        image.save(os.path.join(output_path, output_file))

print("Conversion done!")
```
