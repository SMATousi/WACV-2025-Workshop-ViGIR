```python
# Import libraries
import pandas as pd
import numpy as np
import json
import random
```

```python
# import json

file_path = 'GT_L6_Tol.json'
with open(file_path, 'r') as file:
    data = json.load(file)

print('Number of Annotated GT Images: ', len(data.keys()))
```

```python
# Seed the random generator
seed_number = 2024
random.seed(seed_number)

# Convert the keys to a list and shuffle them randomly
keys = list(data.keys())
random.shuffle(keys)

# Split the keys into two equal non-overlapping parts
split_index = len(keys) // 2
part1_keys = keys[:split_index]
part2_keys = keys[split_index:]

# Create two dictionaries for the split data
part1 = {key: data[key] for key in part1_keys}
part2 = {key: data[key] for key in part2_keys}

print('Number of Validation Images: ', len(part1.keys()))
print('Number of Testing Images: ', len(part2.keys()))
```

```python
# Save the split data into two separate JSON files
with open('Validation_Set.json', 'w') as file1:
    json.dump(part1, file1, indent=4)

with open('Testin.json', 'w') as file2:
    json.dump(part2, file2, indent=4)

print("Data has been split and saved to part1.json and part2.json")
```
